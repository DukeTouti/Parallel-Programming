```python
# ============================================================
# FULL Hadamard Gate Demonstration
# Includes:
#  - Circuit diagram
#  - Statevector
#  - Bloch sphere visualization
#  - Measurement histogram
# ============================================================

%matplotlib inline

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.quantum_info import Statevector
from qiskit.visualization import (
    plot_histogram,
    plot_bloch_multivector
)
from matplotlib import pyplot as plt

print("====================================")
print("STEP 1: Create a 1-qubit circuit")
print("Initial state is |0>")
print("====================================\n")

# Create quantum circuit (1 qubit, 1 classical bit)
qc = QuantumCircuit(1, 1)
print(qc)

# ------------------------------------------------------------
# STEP 2: Apply Hadamard gate
# ------------------------------------------------------------

print("\n====================================")
print("STEP 2: Apply Hadamard gate")
print("H|0> = (|0> + |1>)/√2 → equal superposition")
print("====================================\n")

#qc.x(0)
qc.h(0)

# Draw circuit
qc.draw('mpl')
plt.show()

# ------------------------------------------------------------
# STEP 3: Show statevector BEFORE measurement
# ------------------------------------------------------------

print("\n====================================")
print("STEP 3: Statevector after Hadamard gate")
print("====================================\n")

state = Statevector.from_instruction(qc)
print("Statevector:")
print(state)

# ------------------------------------------------------------
# STEP 4: Bloch sphere visualization
# ------------------------------------------------------------

print("\n====================================")
print("STEP 4: Bloch Sphere Representation")
print("====================================\n")

plot_bloch_multivector(state)
plt.show()

# ------------------------------------------------------------
# STEP 5: Measurement
# ------------------------------------------------------------

print("\n====================================")
print("STEP 5: Measurement")
print("====================================\n")

qc.measure(0, 0)

qc.draw('mpl')
plt.show()

# ------------------------------------------------------------
# STEP 6: Simulation
# ------------------------------------------------------------

print("\n====================================")
print("STEP 6: Run Simulation (1024 shots)")
print("====================================\n")

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print("Measurement Results:")
print(counts)

plot_histogram(counts)
plt.show()

print("\n====================================")
print("FINAL INTERPRETATION:")
print("Hadamard puts |0> into an equal superposition of |0> and |1>")
print("So measuring yields 0 or 1 with ~50% probability each.")
print("Bloch sphere shows the vector on the equator.")
print("====================================")
```
